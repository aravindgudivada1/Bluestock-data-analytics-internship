```python
import pandas as pd
from pathlib import Path

# Path from notebooks folder to raw data
RAW_DATA_DIR = Path("../data/raw")

fund_master = pd.read_csv(RAW_DATA_DIR / "01_fund_master.csv")

print("Dataset shape:", fund_master.shape)
fund_master.head()
```

```python
print("Columns in Fund Master:\n")

for column in fund_master.columns:
    print(column)
```

```python
# Unique fund houses
fund_houses = fund_master["fund_house"].unique()

print("Total unique fund houses:", fund_master["fund_house"].nunique())
print("\nFund Houses:")

for fund_house in fund_houses:
    print(fund_house)
```

```python
# Unique fund categories
categories = fund_master["category"].unique()

print("Total unique categories:", fund_master["category"].nunique())
print("\nCategories:")

for category in categories:
    print(category)
```

```python
# Unique fund sub-categories
sub_categories = fund_master["sub_category"].unique()

print("Total unique sub-categories:", fund_master["sub_category"].nunique())
print("\nSub-Categories:")

for sub_category in sub_categories:
    print(sub_category)
```

```python
# Unique risk categories
risk_categories = fund_master["risk_category"].unique()

print("Total unique risk categories:", fund_master["risk_category"].nunique())
print("\nRisk Categories:")

for risk in risk_categories:
    print(risk)
```

```python
# Load NAV history
nav_history = pd.read_csv(RAW_DATA_DIR / "02_nav_history.csv")

print("Fund Master shape:", fund_master.shape)
print("NAV History shape:", nav_history.shape)

print("\nUnique AMFI codes in Fund Master:",
      fund_master["amfi_code"].nunique())

print("Unique AMFI codes in NAV History:",
      nav_history["amfi_code"].nunique())
```

```python
# Validate that every Fund Master AMFI code exists in NAV History

nav_codes = nav_history[["amfi_code"]].drop_duplicates()

code_validation = fund_master[["amfi_code", "fund_house", "scheme_name"]].merge(
    nav_codes,
    on="amfi_code",
    how="left",
    indicator=True
)

missing_codes = code_validation[code_validation["_merge"] == "left_only"]

print("Total Fund Master codes:", len(code_validation))
print("Matched codes:", (code_validation["_merge"] == "both").sum())
print("Missing codes:", len(missing_codes))

if missing_codes.empty:
    print("\nPASS: All Fund Master AMFI codes exist in NAV History.")
else:
    print("\nWARNING: Some AMFI codes are missing from NAV History:")
    print(missing_codes[["amfi_code", "fund_house", "scheme_name"]])
```
